```python
import pandas as pd

# завантажуємо файл у таблицю (DataFrame)
df = pd.read_csv('index_1.csv')

# показуємо перші 5 рядків
df.head()
```

```python
# подивимося на загальну інформацію про колонки та типи даних
df.info()

```

```python
# перетворюємо колонку 'datetime' у формат дати й часу
df['datetime'] = pd.to_datetime(df['datetime'])

# перетворюємо колонку 'date' у формат дати
df['date'] = pd.to_datetime(df['date'])

# перевіряємо типи даних
df.info()
```

Наступний етап це перевірити чи повні дані. Почнемо з перевірки пустих колонок та пропусків.


```python
# перевіряємо кількість пропусків у кожній колонці

df.isnull().sum()
```

```python
# вибираємо тільки ті замовлення, де в колонці 'card' стоїть пропуск
missing_cards = df[df['card'].isna()]

# показуємо, які типи оплати (cash_type) були у цих 89 замовленнях
missing_cards['cash_type'].value_counts()
```

Перевірили пусті значення у колонці 'card', виявили що 89 пропусків це оплата готівкою. Тому немає даних в колонці 'card'. Натупний етап це заповнити пропуски значенням 'no card'.


```python
# повнюємо порожні клітинки в колонці 'card' текстом 'no card'
df['card'] = df['card'].fillna('no card')

# перевіряємо пропуски ще раз
df.isnull().sum()
```

Переходимо до перевірки повноти даних. А саме до дати, перевіримо часову цілісність.


```python
min_date = df['date'].min()
max_date = df['date'].max()

print("Початок даних:", min_date)
print("Кінець даних:", max_date)
```

```python
# групуємо продажі по місяцях і рахуємо робочі дні та кількість замовлень
df.groupby(df['date'].dt.to_period('M')).agg(
    днів_з_продажами=('date', 'nunique'),
    всього_замовлень=('date', 'count'),
    перший_день=('date', 'min'),
    останній_день=('date', 'max')
)
```

Переходимо до аналізу даних. Exploratory Data Analysis (EDA).

```python
# витягуємо години
df['hour'] = df['datetime'].dt.hour

# рахуємо кількість продажів для кожної години
missing_hour = df['hour'].value_counts()

# виводимо результат на екран

print (missing_hour)
```

```python
# сортуємо по годинах для візуалізації

df['hour'].value_counts().sort_index()
```

# Будуємо граіфк для візуалізації пікових годин продажів кави

```python
import matplotlib.pyplot as plt

# задаємо розмір графіку

plt.figure(figsize=(12, 8))


# просортований підрахунок

hourly_sales = df['hour'].value_counts().sort_index()

# малюємо графік

plt.plot(hourly_sales.index, hourly_sales.values)

# дамо назву графіку та нашим осям

plt.title('Пікові продажі по годинах')
plt.xlabel('Години')
plt.ylabel('Продажі')

# відображаємо графік
plt.show()
```
